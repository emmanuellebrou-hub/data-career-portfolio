```python
configs = {
    "fs.azure.account.auth.type":"CustomAccessToken",
    "fs.azure.account.custom.token.provider.class": spark.conf.get("spark.databricks.passthrough.adls.gen2.tokenProviderClassName")
}

mount_point = "/mnt/globalcsvdata"
source_path = "abfss://globalcsvdata@emmabintechsg.dfs.core.windows.net"

# Check if storage amount already exist
if not any(mount.mountPoint == mount_point for mount in dbutils.fs.mounts()):
    dbutils.fs.mount(
        source = source_path,
        mount_point = mount_point,
        extra_configs = configs
    )
else:
    print(f"Storage amount {mount_point} already exist.")
```

```python
# Get all subfolder of /mnt/globalcsvdata/
for mount in dbutils.fs.ls("/mnt/globalcsvdata"):
    print(mount.path)
    
    # For each folder get all files
    if mount.isDir():
        for sub in dbutils.fs.ls(mount.path):
            print("  " + sub.path)
```
